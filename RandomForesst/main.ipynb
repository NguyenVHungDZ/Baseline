```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import KFold

from src.data_utils import load_data, df_to_tensors, signed_log1p, split_tensors
from src.testing import evaluate_regression
from src.visualization import plot_predicted_vs_observed
from sklearn.ensemble import RandomForestRegressor

SEED = 18
```

```python
data = load_data(str(ROOT / "data" / "features.csv"))
TARGET_COL = "dens"

x_raw, y_raw = df_to_tensors(data, TARGET_COL)

X_train_t, y_train_t, X_test_t, y_test_t = split_tensors(x_raw, y_raw, test_size=0.2, seed=SEED)

X_train = signed_log1p(X_train_t).numpy()
X_test  = signed_log1p(X_test_t).numpy()
y_train = y_train_t.numpy().ravel()
y_test  = y_test_t.numpy().ravel()

print(f"Train: {X_train.shape}  |  Test: {X_test.shape}")
```

```python
# rf = RandomForestRegressor(
#     n_estimators     = 512,
#     max_features     = "sqrt",
#     min_samples_leaf = 2,
#     n_jobs           = -1,
#     random_state     = SEED,
# )

rf = RandomForestRegressor(
    n_estimators     = 512,
    max_features     = "sqrt",
    min_samples_leaf = 1,
    min_samples_split= 2,
    n_jobs           = -1,
    random_state     = SEED,
    max_depth=10,
    bootstrap = False
)

rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test)

print("RANDOM FOREST")
evaluate_regression(y_test, y_pred, verbose=True)

plot_predicted_vs_observed(y_test, y_pred)
```
